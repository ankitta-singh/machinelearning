```python
##2.numpy array attributes
```

```python
import numpy as np
```

```python
arr1 = np.array([[1,2,3], [4,5,6]])
print(arr1)
```

```python
print("shape",arr1.shape)
print("size", arr1.shape)
print("dtype",arr1.dtype)
```

```python
##creating 2 arra-1d and 3d ,and findind it's attribute
```

```python
arr1 = np.array([10,20,30,40,50])
arr3 = np.array([[[1, 2], [3, 4]],[[5, 6], [7, 8]]])
print("1-D Array:")
print(arr1)
print("\n3-D Array:")
print(arr3)
```

```python
print("shape",arr1.shape)
```

```python
print("shape",arr3.shape)
print("size",arr3.size)
```

```python
3.##array initialization method
```

```python
# zeros array
zero_arr = np.zeros((2,3))
print(zero_arr)
```

```python
# one array
one_arr = np.ones((1,3))
print(one_arr)
```

```python
# full array

full_arr = np.full((3,2),2)
print(full_arr)
```

```python
## identiti matrix[all element 0 except diagonal 1]

id_arr = np.eye(3)
print(id_arr)
```

```python
##empty array

print(np.empty(2))
```

```python
##evenly space array

print(np.arange(1,3,1))
```

```python
# specif number of space between a range
print(np.linspace(1,10,4))






      
```
